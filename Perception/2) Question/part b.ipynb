```python
import pandas as pd

# 1. Define Thresholds
RED_THRESHOLD = 0.5
PERSON_THRESHOLD = 0.5

# 2. Define the Baseline Rule
def baseline_decision(row):
    # Rule 1: Pedestrian safety overrides everything
    if row['person_score'] > PERSON_THRESHOLD:
        return pd.Series(['STOP', 'Person detected (score > 0.5)'])
    
    # Rule 2: Evaluate Traffic Light Evidence
    vision_sees_red = row['red_score'] > RED_THRESHOLD
    v2x_says_red = row['v2x_light'] == 'red'
    
    if vision_sees_red and v2x_says_red:
        return pd.Series(['STOP', 'Agreement: Vision and V2X say Red'])
        
    elif vision_sees_red and not v2x_says_red:
        return pd.Series(['SLOW', 'Conflict: Vision sees Red, V2X says Green'])
        
    elif not vision_sees_red and v2x_says_red:
        return pd.Series(['SLOW', 'Conflict: Vision sees Green, V2X says Red'])
        
    else:
        return pd.Series(['GO', 'Agreement: Vision and V2X say Green'])

# 3. Apply rule to the provided dataset
df = pd.read_csv('C:/Users/vinay/OneDrive - Indian Institute of Technology Bhubaneswar/Desktop/sedrica assignment/city/crossing_evidence.csv')
df[['decision', 'reason']] = df.apply(baseline_decision, axis=1)

# 4. View and Save the Output

# Print the relevant columns for the first 20 frames to your terminal
print(df[['episode', 'frame', 'v2x_light', 'decision', 'reason']].head(20))

# Save the final results to a new CSV file in the same folder so you can open it in Excel
output_path = 'C:/Users/vinay/OneDrive - Indian Institute of Technology Bhubaneswar/Desktop/sedrica assignment/city/baseline_results.csv'
df.to_csv(output_path, index=False)

print(f"\nSuccess! Full results saved to: {output_path}")
```
